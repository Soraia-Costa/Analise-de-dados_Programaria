#Featuring engineering - engenharia de recursos,


```python
import pandas as pd
```

```python
dados = pd.read_excel('/content/drive/MyDrive/Nova pasta/planilha_modulo3.xlsx')
```

```python
def preencher_nivel(gestor,nivel):
  if gestor == 1 :
    return "Pessoa Gestora"
  else:
    return nivel
```

```python
dados.columns
```

```python
dados['NOVO_NIVEL'] = dados.apply(lambda x: preencher_nivel(x['GESTOR?'], x['NIVEL']), axis=1)
```

```python
dados['NOVO_NIVEL']
```

```python
dados['NOVO_NIVEL'].value_counts()
```

```python
dados['NIVEL']
```

```python
dados= pd.get_dummies(dados, columns = ['NIVEL'])
```

```python
dados.columns
```

```python
def determinar_geracao(idade) :
  if 39>idade <= 58:
    return "Geracao X"
  elif 29 < idade <= 39 :
    return  "Geracao Millenial"
  elif 13< idade <= 29:
    return "Geracao Z"
  else:
    return "Outra Geracao"
```

```python
dados ["GERACAO"]=dados['IDADE'].apply(determinar_geracao)
```

```python
dados["GERACAO"].value_counts()
```

```python
dados2= pd.read_excel('/content/drive/MyDrive/Nova pasta/aula_Aovivo/Cópia de Planilha_Aula_parte2.xlsx')
```

```python
dados2.head()
```

```python
#funcao merge - mescla tabelas com algum atributo em comum nesse caso foi usado o ID
dados = dados.merge(dados2,on='ID',how='left')

```

```python
dados.columns
```
